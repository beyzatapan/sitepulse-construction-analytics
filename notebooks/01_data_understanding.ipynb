```python
# SitePulse — Data Understanding & Profiling

## Objective

This notebook performs the initial inspection and profiling of the SitePulse construction operations dataset.

The goal is to understand the structure, quality, coverage, and basic characteristics of the data before cleaning, transformation, exploratory analysis, and machine learning.

The analysis focuses on:

- dataset dimensions
- available variables
- data types
- missing values
- duplicate records
- project coverage
- date coverage
- categorical distributions
- numerical summaries
- relational consistency between datasets
```

```python
projects = pd.read_csv(
    RAW_DATA_DIR / "projects.csv"
)

equipment = pd.read_csv(
    RAW_DATA_DIR / "equipment.csv"
)

operations = pd.read_csv(
    RAW_DATA_DIR / "daily_operations.csv"
)

print("Projects:", projects.shape)
print("Equipment:", equipment.shape)
print("Daily operations:", operations.shape)
```

```python
missing_summary = pd.DataFrame(
    {
        "missing_count": operations.isna().sum(),
        "missing_percentage": (
            operations.isna().mean() * 100
        ).round(2),
    }
)

missing_summary = missing_summary[
    missing_summary["missing_count"] > 0
]

missing_summary.sort_values(
    "missing_count",
    ascending=False,
)
```

```python
duplicate_count = operations.duplicated().sum()

duplicate_percentage = (
    duplicate_count
    / len(operations)
    * 100
)

print(f"Duplicate rows: {duplicate_count}")
print(
    f"Duplicate percentage: "
    f"{duplicate_percentage:.2f}%"
)
```

```python
print(
    "Earliest record:",
    operations["date"].min(),
)

print(
    "Latest record:",
    operations["date"].max(),
)
```

```python
event_summary = pd.DataFrame(

    {

        "event": [

            "Fuel anomaly",

            "Material shortage",

            "Equipment breakdown",

        ],

        "count": [

            operations[

                "fuel_anomaly_flag"

            ].sum(),

            operations[

                "material_shortage_flag"

            ].sum(),

            operations[

                "equipment_breakdown_flag"

            ].sum(),

        ],

    }

)

event_summary["percentage"] = (

    event_summary["count"]

    / len(operations)

    * 100

).round(2)

event_summary
```

```python
invalid_operation_projects = set(
    operations["project_id"]
    .dropna()
    .unique()
) - set(
    projects["project_id"]
    .dropna()
    .unique()
)

invalid_operation_projects
```

```python
invalid_equipment_projects = set(
    equipment["project_id"]
    .dropna()
    .unique()
) - set(
    projects["project_id"]
    .dropna()
    .unique()
)

invalid_equipment_projects
```

## Initial Findings

The initial profiling indicates that the SitePulse dataset contains six construction projects and approximately 1,400 daily operational records.

Several data quality issues were identified in the raw dataset:

- missing values are present in workforce, fuel consumption, material delivery, and weather variables
- duplicate operational records are present
- the `date` variable is not yet represented as a datetime data type
- controlled operational events such as fuel anomalies, material shortages, and equipment breakdowns are present

The relational checks indicate that project identifiers used in the equipment and daily operations datasets correspond to valid projects in the project master dataset.

These findings establish the main requirements for the next stage: **data cleaning and validation**.
